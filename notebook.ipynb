# RAG-Based Resume-to-Job Matching System

## Overview
This notebook demonstrates a complete Retrieval Augmented Generation (RAG) system that matches job descriptions to qualified candidates using semantic search and keyword matching.

**Key Features:**
- Smart document chunking (preserves resume sections)
- OpenAI embeddings (1536-dimensional vectors)
- ChromaDB persistent storage (333 embeddings from 30 resumes)
- Hybrid search (semantic 70% + keyword 30%)
- Scoring 0-100 with must-have filtering
- JSON output with reasoning

## Setup

Import required modules and initialize the system.

````python
import json
import os
from job_matcher import run_job_matcher

print("✓ Imports successful")
print(f"✓ Working directory: {os.getcwd()}")
````

## Part 1: System Architecture

### Pipeline Flow
```
Resumes (30) 
    ↓
Smart Chunking (preserve sections)
    ↓
Metadata Extraction (name, years, skills)
    ↓
OpenAI Embeddings (1536-dim vectors)
    ↓
ChromaDB Storage (persistent)
    ↓
Job Description Input
    ↓
Hybrid Search (semantic + keyword)
    ↓
Scoring & Filtering (0-100, must-haves)
    ↓
Top-10 Matches (with reasoning)
```

## Part 2: Test Case 1 - Backend Engineer

````python
jd_backend = """
Senior Backend Engineer - Python FastAPI

We are hiring a Senior Backend Engineer to build scalable microservices.

Requirements:
- 5+ years of backend development experience
- Expert-level Python skills
- Strong experience with FastAPI framework
- PostgreSQL and Redis expertise
- Docker and Kubernetes knowledge
- AWS cloud platform experience
- REST API design and implementation
"""

print("Running job matcher for Backend Engineer JD...\n")
output_backend = run_job_matcher(jd_backend, top_k=5, output_file='notebook_backend_output.json')
````

````python
# Display results
print("\n" + "="*70)
print("TOP 5 MATCHES FOR BACKEND ENGINEER")
print("="*70 + "\n")

for i, match in enumerate(output_backend['top_matches'][:5], 1):
    print(f"{i}. {match['candidate_name']}")
    print(f"   Score: {match['match_score']}/100")
    print(f"   Skills: {', '.join(match['matched_skills'])}")
    print(f"   Reasoning: {match['reasoning']}")
    print()
````
